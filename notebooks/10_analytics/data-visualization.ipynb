# Data Visualization for Gurgaon Properties

This notebook builds analytics-ready datasets and visualizations for the Gurgaon properties pipeline. It covers a geographic map of price and area by sector, a word cloud of property features, and scatter, pie, and distribution plots to support exploration and dashboarding.

```python
# Project paths and data/plotting libraries 
from config import DATA_PROCESSED, DATA_ANALYTICS, MODELS_DIR
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import pickle
from wordcloud import WordCloud
```

## Map: Geographic Visualization of Price and Area by Sector

Load the processed properties dataset and sector coordinates, merge them, and aggregate by sector to produce an interactive scatter map of Gurgaon with point size and color encoding built-up area and price per sqft.

```python
# Load the missing-value-imputed Gurgaon properties dataset for map and later visualizations
df = pd.read_csv(DATA_PROCESSED / 'gurgaon_properties_missing_value_imputation.csv')
```

```python
df.shape
```

```python
# Preview first five rows of the properties dataset
df.head()
```

```python
# Sector-level latitude/longitude for mapping (coordinates stored as formatted strings)
latlong = pd.read_csv(DATA_ANALYTICS / 'latlong.csv')
```

```python
# Display sector-coordinates dataframe
latlong
```

```python
# Parse latitude from coordinate string (e.g. "28.xx° N, 77.xx° E" -> 28.xx)
latlong['latitude'] = latlong['coordinates'].str.split(',').str.get(0).str.split('°').str.get(0).astype('float')
```

```python
# Parse longitude from coordinate string
latlong['longitude'] = latlong['coordinates'].str.split(',').str.get(1).str.split('°').str.get(0).astype('float')
```

```python
latlong.head()
```

```python
# Attach sector coordinates to each property row for map plotting
new_df = df.merge(latlong, on='sector')
```

```python
# Preview properties with sector coordinates attached
new_df.head()
```

```python
# Column names of the merged properties-with-coordinates dataframe
new_df.columns
```

```python
# Aggregate by sector: mean price, price_per_sqft, built_up_area, and single lat/lon per sector for the map
group_df = (
    new_df.groupby('sector')
    [['price','price_per_sqft','built_up_area','latitude','longitude']]
    .mean()
)
```

```python
# Sector-level aggregates (mean price, price_per_sqft, built_up_area, lat, lon)
group_df
```

```python
# Interactive map: color = price_per_sqft, size = built_up_area; sector labels on hover
fig = px.scatter_map(group_df, lat="latitude", lon="longitude", color="price_per_sqft", size='built_up_area',
                  color_continuous_scale=px.colors.cyclical.IceFire, zoom=10,
                  map_style="open-street-map",text=group_df.index)
fig.show()
```

```python
# Export merged property–coordinates data for dashboards or further analytics
new_df.to_csv(DATA_ANALYTICS / 'data_viz1.csv', index=False)
```

## Word Cloud: Property Features from Listings

Use the raw Gurgaon properties dataset to extract the `features` field (list of amenities per listing), flatten them into a single corpus, and generate a word cloud plus a serialized feature text.

```python
# Raw properties file contains the 'features' column (amenity lists) needed for the word cloud
df1 = pd.read_csv(DATA_PROCESSED / 'gurgaon_properties.csv')
```

```python
df1.head()
```

```python
# Align raw and processed data by index and keep features plus sector for word cloud
wordcloud_df = df1.merge(df, left_index=True, right_index=True)[['features','sector']]
```

```python
# Preview features and sector for word cloud pipeline
wordcloud_df.head()
```

```python
#pickle.dump(wordcloud_df, open('wordcloud_df.pkl', 'wb'))
```

```python
# Parse 'features' and flatten into a single list of amenity names for the word cloud
import ast
main = []
for item in wordcloud_df['features'].dropna().apply(ast.literal_eval):
    main.extend(item)
```

```python
# Flattened list of all amenity/feature names
main
```

```python
# Single space-separated string of all feature tokens for WordCloud and optional export
feature_text = ' '.join(main)
```

```python
feature_text
```

```python
# Save feature text
import pickle
pickle.dump(feature_text, open(MODELS_DIR / 'feature_text.pkl', 'wb'))
```

```python
plt.rcParams["font.family"] = "Arial"

wordcloud = WordCloud(width = 800, height = 800, 
                      background_color ='white', 
                      stopwords = set(['s']),  
                      min_font_size = 10).generate(feature_text)

plt.figure(figsize = (8, 8), facecolor = None) 
plt.imshow(wordcloud, interpolation='bilinear') 
plt.axis("off") 
plt.tight_layout(pad = 0) 
plt.show()
```

## Scatter Plot, Pie Chart, and Distribution Visualizations

Build scatter (built-up area vs price, colored by bedroom count), pie (bedroom distribution), boxplot (price by bedroom), and distribution plots (price by property type) for exploratory analysis and reporting.

```python
data = dict(
    names=["A", "B", "C", "D", "E", "F"],
    parents=["", "", "", "A", "A", "C"],
    values=[10, 20, 30, 40, 50, 60],
)

fig = px.sunburst(
    df1,
    names='property_type',
    values='price_per_sqft',
    parents='bedRoom',
    title="Sample Sunburst Chart"
)
fig.show()
```

```python
# Scatter: built_up_area vs price, colored by bedroom count
fig = px.scatter(df, x="built_up_area", y="price", color="bedRoom", title="Area Vs Price")

# Show the plot
fig.show()
```

```python
# Pie chart: distribution of listings by bedroom count
fig = px.pie(df, names='bedRoom', title='Distribution by Bedroom Count')

# Show the plot
fig.show()
```

```python
# Restrict to 1–4 BHK for clearer boxplot; price (in crore) by bedroom count
temp_df = df[df['bedRoom'] <= 4]
# Side-by-side boxplots of price (crore) by bedroom count
fig = px.box(temp_df, x='bedRoom', y='price', title='BHK Price Range')

# Show the plot
fig.show()
```

```python
# Price distribution by property type: house vs flat
sns.kdeplot(df[df['property_type'] == 'house']['price'], label='House', fill=True)
sns.kdeplot(df[df['property_type'] == 'flat']['price'], label='Flat', fill=True)
plt.legend()
plt.show()
```
